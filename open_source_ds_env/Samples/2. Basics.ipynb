```python
#STEP 1: Import Tools
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#STEP 2: Load or Create data
data = {
    'Years_Experience': [1, 2, 3, 5, 8, 10],
    'Salary_USD': [45000, 50000, 60000, 75000, 95000, 110000]
}

#df = pd.DataFrame(data)                                             # Load data from Program
df = pd.read_csv('employee_data.csv')                        # Load Data from CSV File
print(f"\n ----- Raw Dataset -----")
print(df)
                                                                     # Load data from Url
#url="https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/breast-cancer-wisconsin.data"
#name=["id","clump_thickness","uniform_cell_size","uniform_cell_shape","marginal_adhesion","single_epithelial_size","bare_nuclei","bland_chromatin","normal_nucleoli","mitoses","class"]
#df=pd.read_csv(url,names=name)
```

```python
#STEP 3: Clean/Calculate/Analyze data
df_cleaned = df.dropna(subset=['Department'])                        # CLEAN MISSING DATA - Remove employees who doesn't have department

average_salary = df_cleaned['Salary_USD'].mean()                     # Calculate Mean of Salary_USD
print(f"\nAverage Company Salary: ${average_salary:.2f}")

department_salaries = df_cleaned.groupby('Department')['Salary_USD'].mean()  # Group by Dept & then Calculate Mean of Salary_USD
print(f"\n--- Average Salary by Department ---")
print("\n",department_salaries)

sorted_df = df_cleaned.sort_values(by='Salary_USD', ascending=False) # Sort by Salary_USD
print("\n",sorted_df)

condition_dept = (sorted_df['Department'] == 'Data')                        # Filter by Department = Data
condition_salary = (sorted_df['Salary_USD'] > 60000)                        # Filter by Salary_USD > 60000
filtered_df = sorted_df[condition_dept & condition_salary]                  # New Filtered Dataframe
print("\n",filtered_df)
```

```python
#STEP 4: Visualize the results
plt.scatter(df['Years_Experience'], df['Salary_USD'], color='blue')
plt.title('Salary vs. Experience')
plt.xlabel('Years of Experience')
plt.ylabel('Salary ($)')
plt.show()
```
